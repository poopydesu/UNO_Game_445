```python
# Import everything defined in this Logic Python library
from logic import *
```

```python
wumpus_kb = PropKB()   # A KB for a Wumpus game
P11, P12, P21, P22, P31, B11, B21 = expr('P11, P12, P21, P22, P31, B11, B21')
wumpus_kb.tell(~P11)   # Observation: No Pit in [1,1]
wumpus_kb.tell(~B11)   # Observation: No Breeze at [1,1]
wumpus_kb.tell(B21)    # Observation: Breeze at [2,1]
# Game rules: A tile is breezy if and only if there is a pit in a neighboring square. 
wumpus_kb.tell(B11 | '<=>' | ((P12 | P21)))
wumpus_kb.tell(B21 | '<=>' | ((P11 | P22 | P31))) 
```

```python
# We can check the clauses stored in a KB by accessing its clauses variable
wumpus_kb.clauses
```

```python
tt_entails(P & Q, Q)   # P & Q is True only when both P and Q are True. So (P & Q) => Q is True.
```

```python
tt_entails(P | Q, Q)   # If we know that P | Q is True, we cannot infer the truth values of P and Q. 
                       # Hence, (P | Q) => Q is False.
tt_entails(P | Q, P)   # and similar is (P | Q) => P
```

```python
(A, B, C, D, E, F, G) = symbols('A, B, C, D, E, F, G')
tt_entails(A & (B | C) & D & E & ~(F | G), A & D & E & ~F & ~G)
```
