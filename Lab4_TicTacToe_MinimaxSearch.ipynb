```python
import math, time
class TicTacToe:
    def __init__(self, useAlphaBeta=True):
        """Initialize the Tic Tac Toe game"""
        self.state=[[' ' for i in range(3)] for j in range(3)]   # default state (empty board)
        self.turn = 'X'                                          # player turn: AI plays X first
        self.useAlphaBeta = useAlphaBeta                         # flag to use Alpha Beta pruning
        self.scores = {'X': 1, 'tie': 0, 'O': -1}                # terminal score
        
    def draw(self):
        """Print the game board from the given state"""
        for i in range(3):
            print('+---+---+---+')
            print('|',self.state[i][0],'|',self.state[i][1],'|',self.state[i][2],'|')
        print('+---+---+---+')
        
    def validMove(self, px, py):
        """Return if making a move at the position px, py is valid or not"""
        if px < 0 or px > 2 or py < 0 or py > 2 or self.state[px][py] != ' ':
            return False
        return True
    
    def checkWin(self):
        """Check the current state of the game for any winner, tie, or not finished"""
        # check horizontal
        for i in range(3):
            if (self.state[i][0] != ' ' and self.state[i][0] == self.state[i][1] 
                and self.state[i][1] == self.state[i][2]):
                return self.state[i][0]
        # check vertical
        for j in range(3):
            if (self.state[0][j] != ' ' and self.state[0][j] == self.state[1][j] 
                and self.state[1][j] == self.state[2][j]):
                return self.state[0][j]
        # check diagonal
        if (self.state[0][0] != ' ' and self.state[0][0] == self.state[1][1] 
            and self.state[1][1] == self.state[2][2]):
            return self.state[0][0]
        if (self.state[0][2] != ' ' and self.state[0][2] == self.state[1][1] 
            and self.state[1][1] == self.state[2][0]):
            return self.state[0][2]
        # count empty slots to determine tie or unfinished game
        for i in range(3):
            for j in range(3):
                if self.state[i][j] == ' ':
                    return None   # game is not finished yet
        # No empty slots means the game is a tie
        return 'tie'

    #TODO: Complete the missing code of the minimax algorithm below
    def minimax(self, state, depth, isMax, alpha=-math.inf, beta=math.inf):
        """Implement recursive Minimax algorithm with optional alpha-beta to return the best score"""
        winner = self.checkWin() # check if there is a winner AI(X) or human (O)
        if winner:  # if the game is in the terminal state, return the corresponding score
            return self.scores[winner]
        if isMax:     #Max Node
            #TODO: Initialze bestScore with -math.inf
            bestScore = -math.inf
            #TODO: Loop on rows and columns to search for an empty tile [i, j]
            for i in range(3):
                for j in range(3):
                    if state[i][j] == ' ':
                #TODO: Assume the state[i][j] = 'X'
                        state[i][j] = 'X';
                #TODO: Estimate the score by recursively calling minimax 
                        score = self.minimax(state, depth+1, False, alpha, beta)
                #         with such state, depth+1, isMax=false, and current alpha beta
                #TODO: Reset state[i][j] back to ' '
                        state[i][j] = ' '
                #TODO: Set bestScore = max of score and bestScore
                        bestScore = max(bestScore, score)
                #TODO: if useAlphaBeta is true:
                        if self.useAlphaBeta:
                    #TODO: Set alpha = max of alpha and score
                            alpha = max(alpha, score)
                    #TODO: if bestScore >= beta, ret bestScore
                            if bestScore >= beta:
                                return bestScore
            #TODO: return bestScore
            return bestScore
        else:     #Min Node
            #TODO: Initialize bestScore = math.inf
            bestScore = math.inf
            #TODO: Loop on rows and columns to search for an empty tile [i, j]
            for i in range(3):
                for j in range(3):
                    if state[i][j] == ' ':
                #TODO: Assume the state[i][j] = 'O'
                        state[i][j] = 'O';
                #TODO: Estimate the score by recursively calling minimax 
                        score = self.minimax(state, depth+1, True, alpha, beta)
                #         with such state, depth+1, isMax=True, and current alpha beta
                #TODO: Reset state[i][j] back to ' '
                        state[i][j] = ' '
                #TODO: Set bestScore = min of score and bestScore
                        bestScore = min(score, bestScore)
                #TODO: if useAlphaBeta is True:
                        if self.useAlphaBeta:
                    #TODO: Set beta = min of beta and score
                            beta = min(beta, score)
                    #TODO: if bestScore <= alpha, return bestScore
                            if bestScore <= alpha:
                                return bestScore
            #TODO: return bestScore
            return bestScore

    def bestMove(self, isMax=True):
        """Find a best move depends on the current player is max or min 
                (depending on the isMax variable)"""
        if isMax:   # Maximizing player
            bestScore = -math.inf
            bestMove = (0, 0)
            for i in range(3):
                for j in range(3):
                    if self.state[i][j] == ' ':   #tile (i,j) is not clicked before
                        self.state[i][j] = 'X'
                        score = self.minimax(self.state, 0, False)
                        self.state[i][j] = ' '
                        if score > bestScore:
                            bestScore = score
                            bestMove = (i, j)
            return bestMove
        else:
            bestScore = math.inf
            bestMove = (0, 0)
            for i in range(3):
                for j in range(3):
                    if self.state[i][j] == ' ':   #tile (i,j) is not clicked before
                        self.state[i][j] = 'O'
                        score = self.minimax(self.state, 0, True)
                        self.state[i][j] = ' '
                        if score < bestScore:
                            bestScore = score
                            bestMove = (i, j)
            return bestMove
    
    def play(self):
        """Main loop to play the game"""
        while True:
            self.draw()    # first draw the tic tac toe board
            self.result = self.checkWin()  # check if the game is finished or not
            if self.result != None:
                if self.result == 'X':
                    print('X win!')
                elif self.result == 'O':
                    print('O win!')
                elif self.result == 'tie':
                    print('Tie!')
                return
            
            if self.turn == 'O':    # Human
                start = time.time()
                move = self.bestMove(False)
                end = time.time()
                print('Player turn (O): Evaluation time: {}s'.format(round(end - start, 7)))
                print('Recommended move: X = {}, Y = {}'.format(move[0], move[1]))
                while True:
                    # As this human turn, ask he/she for which grid cell to play O
                    px, py = map(int, input('Enter the X, Y coordinate to place O: ').split())
                    if self.validMove(px, py):
                        self.state[px][py] = 'O'
                        self.turn = 'O' if self.turn == 'X' else 'X'   # switch the turn to AI
                        break
                    else:
                        print('The move is not valid! Try again.')
            else:    # AI turn - play X
                print('AI turn (X):')
                move = self.bestMove()
                self.state[move[0]][move[1]] = 'X'
                self.turn = 'O' if self.turn == 'X' else 'X' # switch the turn to human
```
